```python
import json
import numpy as np
import pandas as pd

# Load temperature data from CSV file
df = pd.read_csv(r'C:\Users\Monthawat\Downloads\temperature_data_2026-04-22T11_58_56.517Z.csv')

# Extract kelvin temperatures (convert from Celsius to Kelvin if needed)
temperatures = df['Thermo Ch 5'].values

# Convert to numpy array
temperatures = np.array(temperatures)

# Calculate mean and standard deviation
mean_temp = np.mean(temperatures)
std_temp = np.std(temperatures)

# Calculate Signal to Noise Ratio (SNR) as mean / std
snr = mean_temp / std_temp

# Calculate SNR in dB (assuming voltage ratio, so 20*log10)
snr_db = 20 * np.log10(snr)

print(f"Mean Temperature: {mean_temp:.4f} C")
print(f"Standard Deviation: {std_temp:.6f} C")
print(f"Signal to Noise Ratio: {snr:.2f}")
print(f"Signal to Noise Ratio (dB): {snr_db:.2f} dB")
print(f"Inverse Signal to Noise Ratio: {1/snr:.10f}")
```

```python
import json

# Create channel data
channel_data = {
    "channel": 5,
    "temperatures": temperatureValues[:200][::-1].tolist(),
    "sensors": resistanceValues[:200][::-1].tolist()
}

# Write to JSON file
with open('channel_data.json', 'w') as f:
    json.dump(channel_data, f, indent=2)

print("JSON file created: channel_data.json")
print(f"Channel: {channel_data['channel']}")
print(f"Temperature data points: {len(channel_data['temperatures'])}")
print(f"Sensor data points: {len(channel_data['sensors'])}")
```

```python
# Create .340 calibration file
with open('calibration.340', 'w') as f:
    # Write header
    f.write("Sensor Model: NTC10K3950\n")
    f.write("Serial Number: 67676767\n")
    f.write("Data Format: 4\n")
    f.write("SetPoint Limit: 432.15\n")
    f.write("Temperature coefficient: 1\n")
    f.write("Number of Breakpoints: 50\n")
    f.write("\n")
    f.write("No.\tUnits\tTemperature (K)\n")
    
    
    # Write data rows
    for i in range(len(temperatureValues[30:80])):
        row_num = i + 1
        resistance = np.log10(resistanceValues[30:80][::-1][i])
        temperature = temperatureValues[30:80][::-1][i]
        f.write(f"{row_num}\t{resistance:.5f}\t{temperature:.3f}\n")

print("Calibration file created: calibration.340")
print(f"Total breakpoints: {len(temperatureValues[30:80])}")
```
